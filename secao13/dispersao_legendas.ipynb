# Dispersão

```python
import matplotlib.pyplot as plt 
import pandas as pd
```

## Dispersão simples

```python
base = pd.read_csv('data/raw/co2.csv')
base
```

Um loop para plotar as variáveis por tratamento.

```python
x = base['conc']
y = base['uptake']
print(x)
print(y)
```

```python
# Retorna valores únicos do tratamento
unicos = list(set(base['Treatment']))
unicos
```

O loop abaixo, para cada tratamento, retorn um scatterplot que irá se sobrepor para os vários tratamentos, adicionando legendas

```python
for i in range(len(unicos)):
    indice = base['Treatment'] == unicos[i]
    plt.scatter(x[indice], y[indice], label=unicos[i])
plt.legend(loc='lower right')
```

## Multiplos gráficos de dispersão

```python
base = pd.read_csv('data/raw/trees.csv')
base.head()
```

## Circunferência por volume

```python
plt.scatter(base['Girth'], base['Volume'])
```

## Circunferência por altura

```python
plt.scatter(base['Girth'], base['Height'])
```

```python
plt.scatter(base['Height'], base['Girth'])
```

## Todos os gráficos em um só plot

```python
plt.figure(1)
plt.subplot(2, 2, 1)
plt.scatter(base['Girth'], base['Volume'])
plt.subplot(2, 2, 2)
plt.scatter(base['Girth'], base['Height'])
plt.subplot(2, 2, 3)
plt.scatter(base['Height'], base['Girth'])
plt.subplot(2, 2, 4)
plt.hist(base['Volume'])
```
