# Histogramas

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

## Leitura

```python
base = pd.read_csv("data/raw/dados.csv", sep=';')
```

```python
base.head()
```

## Criação do histograma

```python
h = np.histogram(base.iloc[:,2], bins=6)
```

## Visualização

```python
plt.figure()
plt.hist(base.iloc[:,1], bins=6)
plt.title('Árvores')
plt.xlabel('Altura')
plt.ylabel('Frequência')
plt.show()
```
